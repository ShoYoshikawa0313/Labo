```python
import os
import datetime
from optimizer import Random_Optimizer
```

```python
class Args:
    def __init__(self, model = "Model1", composition_file = "composition.yaml", root_output_dir = "results", dir_name = "", seed = 0, processes = 4, immigration_type = "", immigration_freq = 1):
        self.model = model
        self.composition_file = composition_file
        self.root_output_dir = root_output_dir
        self.dir_name = dir_name
        self.seed = seed
        self.processes = processes
        self.immigration_type = immigration_type
        self.immigration_freq = immigration_freq
```

```python
process_composits = [
    #Args(model="Model6",dir_name="no_immigration",immigration_type=""),
    #Args(model="Model6",dir_name="random",immigration_type="random"),
    #Args(model="Model6",dir_name="diversity",immigration_type="diversity"),
    #Args(model="Model6",dir_name="score",immigration_type="score"),
    Args(model="Model6",dir_name="novelty",immigration_type="novelty"),
    Args(model="Model6",dir_name="cluster_novelty",immigration_type="cluster_novelty"),
]
```

```python
for process_composit in process_composits:
    if not os.path.exists(process_composit.root_output_dir):
        os.mkdir(process_composit.root_output_dir)

    if process_composit.dir_name == "":
        dt = datetime.datetime.now() + datetime.timedelta(hours=9)
        process_composit.root_output_dir = os.path.join(process_composit.root_output_dir,dt.strftime('%m-%d_%H:%M'))
    else:
        process_composit.root_output_dir = os.path.join(process_composit.root_output_dir,process_composit.dir_name)

    if not os.path.exists(process_composit.root_output_dir):
        os.mkdir(process_composit.root_output_dir)

    optimizer = Random_Optimizer(args=process_composit)
    optimizer.optimize()

    print(f"process {process_composit.immigration_type} finished !!")
```

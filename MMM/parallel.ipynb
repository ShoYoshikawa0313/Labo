```python
import multiprocessing
import time

def worker_func(x):
    print(f"start {x}")
    time.sleep(x)
    print(f"finish {x}")
    return x * x

# withステートメントを使用してPoolを作成
# プロセス数を指定しない場合、os.cpu_count()の値が使われます
if __name__ == '__main__':
    with multiprocessing.Pool(processes=4) as pool:
        # map()を使用して複数のタスクを投入し、結果を順序通りに取得
        data = [8,7,6,5,4,3,2,1]
        results = pool.map(worker_func, data)
        
        print(f"元のデータ: {data}")
        print(f"結果: {results}") 
        # 出力例: [1, 4, 9, 16, 25, 36, 49, 64]
    
    # withブロックを抜けると、pool.close() と pool.join() が自動的に実行され、
    # ワーカープロセスが適切に終了します。
```
